Avant de passer aux différents étapes de la préparation des données, nous avons prépare un notebook documenté ou a préparé un vocabulaire concernant la réglementation bancaires qui pourrait nous aider à labeliser notre jeu de données pour qui'il soit préparé pour la classification.

Ceci est le résultat du travail éffectué dans le notebook. En utilisant le NLP, on préparé des patterns en se basant sur des documents de la réglemenation bancaire et de la réglementation des changes, des définitions sur sites web de certaines banques comme la notre (Attijari). On va utiliser ces patterns pour filtrer et labeliser notre jeu de données.

Pour plus d'information, vous pouvez le consulter: [ce notebook](https://www.kaggle.com/votre_nom/votre_notebook).

```python
patterns = {
    'banking': r'\b(compte|professionnel|devise|convertible)\b.*?\b(compte|professionnel|devise|convertible)\b',
    'transfers': r'\b(transfert|frais|circulaire|article|courant|change|service|titre|condition|fixer)\b', 
    'accounts': r'\b(compte|devise|professionnel|autre|titulaire|tunisie|pouvoir|central|circulaire|tout)\b',
    'products': r'\b(carte|international|corp|transaction|technologique|allocation|annuel|exclusivement|permettre|site|concurrence)\b',
    'services': r'\b(droit|plafond|devise|autre|cours|service|bien|marches|convertible|contrat)\b',
    'payments': r'\b(article|titre|circulaire|transfert|carte|pouvoir|paiement|service|international|tout)\b'
}
```

Cette fonction sera applique sur notre dataset en utilisant les patternes qu'on a défini. Cette fonction cherche des mots dans chacune des colonnes (LIB_CPTE, PRODUIT et DEV_CPTE) et vérifie si elles correspondent à celles des patternes.

```python
def check_rule_compliance(row):
    
    for feature, pattern in patterns.items():
        if re.search(pattern, str(row[feature]), re.IGNORECASE):
            return 'Non-Compliant'  # If any pattern matches, mark as non-compliant
    return 'Compliant'

# Apply the compliance check to each transaction
td['RULE_COMPLIANCE'] = td.apply(check_rule_compliance, axis=1)
```

Un autre filtre sera appliqué sur notre data qui correspond a montant annuel total que le teneur de la carte technologique internationale peut dépenser. On a trouvé ça aussi dans le document de la réglementation des changes, mais on ne peut définir ce filtre dans un pattern en utilisant puisque celui ci est basé sur des chiffres.

Le montant ne doit pas dépasser 1000 TND par année alors on extracté l'année du champ **DATE_TRX** et on a aggrégré le montant par le teneur de la carte (**REF_UNIQUE**) et l'année extractée pour détecter ce cas de non conformité.

```python
def check_annual_limit(transactions, card_type='Carte Technologique Internationale', annual_limit=1000):
    # Filter transactions by card type
    tech_card_transactions = transactions[(transactions['PRODUIT'] == card_type) & (transactions['RULE_COMPLIANCE'] == 'Compliant')]


    # Group by cardholder and year to sum the transactions
    tech_card_transactions['YEAR'] = tech_card_transactions['DATE_TRX'].dt.year
    annual_spending = tech_card_transactions.groupby(['REF_UNIQUE', 'YEAR'])['AMOUNT'].sum().reset_index()

    # Check for non-compliance based on annual limit
    annual_spending['IS_COMPLIANT'] = annual_spending['AMOUNT'] <= annual_limit

    # Identify non-compliant cardholders
    non_compliant_cardholders = annual_spending[annual_spending['IS_COMPLIANT'] == False]['REF_UNIQUE'].unique()

    # Mark non-compliant transactions in the original dataframe
    transactions.loc[(transactions['REF_UNIQUE'].isin(non_compliant_cardholders)) & 
                     (transactions['PRODUIT'] == card_type), 'ANNUAL_LIMIT_COMPLIANCE'] = 'Non-Compliant'
    
    return transactions
```

```python
td['ANNUAL_LIMIT_COMPLIANCE'] = 'Compliant'

# Apply the annual limit compliance check
td = check_annual_limit(td)

# Final compliance check
td['FINAL_COMPLIANCE'] = td.apply(lambda row: 'Non-Compliant' if row['RULE_COMPLIANCE'] == 'Non-Compliant' or row['ANNUAL_LIMIT_COMPLIANCE'] == 'Non-Compliant' else 'Compliant', axis=1)
```

```python
def plot_numerical_features_boxplots(data, columns_list, title):
    num_features = len(columns_list)
    cols = 3
    rows = (num_features + cols - 1) // cols

    sns.set_style('darkgrid')
    fig, axs = plt.subplots(rows, cols, figsize=(18, 7 * rows), sharey=True)
    fig.suptitle(title, fontsize=25, y=1)

    plt.subplots_adjust(hspace=0.5)

    axs = axs.flatten()

    outliers_df = pd.DataFrame(columns=['Column', 'Outlier_index', 'Outlier_values'])

    for i, col in enumerate(columns_list):
        sns.boxplot(x=data[col], color='#F2F4F4', ax=axs[i])
        axs[i].set_title(f'{col} (skewness: {round(float(data[col].skew()), 2)})', fontsize=12)

        #----------------------------
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        outliers = ((data[col] < (Q1 - 1.5 * IQR)) | (data[col] > (Q3 + 1.5 * IQR)))
        outliers_index = data[outliers].index.tolist()
        outliers_values = data[col][outliers].tolist()

    
        outliers_df = pd.concat([outliers_df, pd.DataFrame({'Column': [col], 'Outlier_index': [outliers_index], 'Outlier_values': [outliers_values]})], ignore_index=True)

        axs[i].plot([], [], 'ro', alpha=0.5, label=f'Outliers: {outliers.sum()}')
        axs[i].legend(loc='upper right', fontsize=10)

    for i in range(num_features, rows * cols):
        axs[i].axis('off')

    plt.tight_layout(rect=[0, 0, 1, 0.97])  # Adjust the layout
    return outliers_df

outliers_df_result = plot_numerical_features_boxplots(data=td_numerical, columns_list=td_numerical.columns, title='Boxplots for Outliers')
```

#### Variance Thresholding

La **variance** est la mesure du changement dans une fonctionnalité donnée. Par exemple, si tous les échantillons d’une caractéristique ont les mêmes valeurs, cela signifie que la variance de cette caractéristique est nulle. Il est essentiel de comprendre qu’une colonne qui n’a pas suffisamment de variance est aussi bonne qu’une colonne avec toutes les valeurs « nan » ou manquantes. S’il n’y a aucun changement dans la fonctionnalité, il est impossible d’en déduire un modèle. Nous vérifions donc la variance et éliminons toute fonctionnalité qui présente une variation faible ou inexistante.

Les seuils de variance peuvent être un bon moyen d'éliminer des caractéristiques dans des ensembles de données, mais dans les cas où il existe des classes minoritaires (par exemple, 5 % de 0 et 95 % de 1), même de bonnes caractéristiques peuvent avoir une très faible variance et finir par être de très bons prédicteurs. Soyez donc avisé : gardez le ratio cible à l’esprit et utilisez des méthodes de corrélation avant d’éliminer des fonctionnalités uniquement basées sur la variance.

- Nous calculons la variance de chaque caractéristique.
- Nous identifions les caractéristiques avec une très faible variance qui peuvent ne pas fournir beaucoup d'informations et peuvent être supprimées.

```python
warnings.filterwarnings('ignore', category=FutureWarning)
threshold_value = 0.01
selector = VarianceThreshold(threshold=threshold_value)
selector.fit(td_numerical)

feature_variances = selector.variances_

# Calculate percentage variance
total_variance = sum(feature_variances)
feature_variances_percentage = (feature_variances / total_variance) * 100

# Create a DataFrame for better plotting with Seaborn
import pandas as pd
feature_df = pd.DataFrame({
    'Feature': td_numerical.columns,
    'Variance (%)': feature_variances_percentage
})

# Plot using Seaborn barplot
plt.figure(figsize=(8, 4))
sns.barplot(x='Feature', y='Variance (%)', data=feature_df, color='#eb563d')
plt.axhline(y=threshold_value * 100, color='red', linestyle='--', label=f'Seuil ({threshold_value * 100:.2f}%)')
plt.xlabel('Caractéristiques')
plt.ylabel('Variance (%)')
plt.title('Variances des caractéristiques avec seuil')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Summary of features below threshold
below_threshold_features = feature_df[feature_df['Variance (%)'] < threshold_value * 100]
print(f"Caractéristique avec la variance en dessous du seuil ({threshold_value * 100:.2f}%):")
print(below_threshold_features)
```

```python
# Identify columns to drop based on variance threshold
columns_to_drop_variance = td_numerical.columns[~selector.get_support()]
print("Les colonnes qui peuvent être éliminées en se basant sur le seuil de la variance:", columns_to_drop_variance)
```

#### Scaling

On mis notre jeu de données à l'échelle pour qu'il sera prêt à la modélisation. On a utilisé la classe normalize de pandas.

```python
def normalizing(data):
    # Exclude encoded columns from normalization
    columns_to_normalize = [col for col in data.columns ]

    for col in columns_to_normalize:
        data[col] = normalize([np.array(data[col])]).reshape(-1)

    return data
```

Au cours de cette étape, nous divisons généralement notre ensemble de données en ensembles d'entrainement et de test, en nous assurant que les données sont bien réparties entre ces ensembles.
```python
# Perform a stratified split to maintain the proportion of the target variable in both sets
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, stratify=y, random_state=42)
```

```python
def rf_model(train_data, test_data):
    model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
    model.fit(train_data, y_train)
    preds = model.predict(test_data)
    return preds, model
```
